```python
def loan_emi(amount, duration):
  emi = amount / duration
  print("Your monthly emi to this {} months is ${}".format(duration,emi))
```

```python
loan_emi(1000000, 10)
```

```python
def loan_amount(amount, duration, down_payment = 0):
  loan_amount = amount - down_payment
  emi = loan_amount/duration
  print("The emi per month: {}".format(emi))
```

```python
loan_amount(100000, 12, 10000)
```
